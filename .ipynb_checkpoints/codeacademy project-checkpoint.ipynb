```python
from langchain_google_genai import GoogleGenerativeAI
import os
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")

# Try different Gemini integration
llm = GoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=api_key,
    temperature=0.7
)

print("Gemini LLM initialized with LangChain!")
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
question_prompt = PromptTemplate.from_template("Generate a beginner-level question about {topic}")
```

```python
answer_prompt = PromptTemplate.from_template("Answer with details explanation from {question}")
```

```python
question_chain = question_prompt | llm | StrOutputParser()
answer_chain = answer_prompt | llm | StrOutputParser()

print("Chains created successfully!")
```

```python
def run_quiz_generator(topic):
    """Run question and answer chains sequentially"""
    # First chain: generate question
    question = question_chain.invoke({"topic": topic})
    
    # Second chain: generate answer from question
    answer = answer_chain.invoke({"question": question})
    
    return {
        "topic": topic,
        "question": question,
        "answer": answer
    }

print("Quiz generator function created!")
```

```python
# Test the quiz generator
result = run_quiz_generator("Инвестиция в недвижимость в Рас Аль Хайме")

print("Topic:", result["topic"])
print("\nQuestion:", result["question"])
print("\nAnswer:", result["answer"])
```
